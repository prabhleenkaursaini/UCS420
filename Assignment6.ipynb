Q1. Consider the following sensor readings:

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])


A sensor has an average measurement error of +2°C.



Perform the following without using a for loop:


a. Add 2°C to every reading using vectorization.


```python
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
corrected_temperature = temperature+2
print("Original temperature:", temperature)
print("Temperature after adding 2°C:", corrected_temperature)
```

b. Convert all temperatures from Celsius to Fahrenheit using: F = 9/5 C + 32


```python
fahrenheit = 9/5*corrected_temperature+32
print("Temperature in Fahrenheit:", fahrenheit)
```

c. Identify all readings greater than 32°C using Boolean indexing.


```python
readings=temperature[temperature>32]
print("Readings greater than 32°C:", readings)
```

d. Count how many readings exceed 32°C.


```python
count = np.sum(temperature>32)
print("Number of readings greater than 32°C:", count)
```

```python
print("Vectorization performs operations on the complete array at once,")
print("while Boolean indexing selects required values directly without a for loop.")
```

Q2. The following matrix represents the number of daily steps recorded by four users over three days:


steps = np.array([


[5000, 6200, 7100],


[8000, 7500, 9000],


[4500, 5100, 4800],


[9000, 8500, 9500]])


Perform the following:


a. Calculate the total steps recorded.


```python
steps = np.array([[5000, 6200, 7100],[8000, 7500, 9000],[4500, 5100, 4800],[9000, 8500, 9500]])
total_steps = np.sum(steps)
print("Total steps:", total_steps)
```

b. Calculate the mean number of steps.


```python
mean_steps = np.mean(steps)
print("Mean number of steps:", mean_steps)
```

c. Find the maximum and minimum values.


```python
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))
```

d. Calculate the total steps for each day using axis=0.


```python
day_total = np.sum(steps, axis=0)
print("Total steps for each day:", day_total)
```

e. Calculate the total steps for each user using axis=1.


```python
user_total = np.sum(steps, axis=1)
print("Total steps for each user:", user_total)
```

f. Find the user/day position containing the maximum number of steps using argmax().


```python
max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("User/day position:", max_position)
print("Maximum steps:", steps[max_position])
```

Q4. A simple cognitive-assistive system wants to predict a user's assistance score based on three factors:


• Sleep hours


• Activity level


• Stress level


Consider:


X = np.array([


[6, 70, 3],


[5, 50, 6],


[8, 80, 2],


[4, 30, 8]


])


y = np.array([40, 65, 30, 85])


Here, each row represents one user and each column represents a feature.


Perform the following:


a. Display the shape and dimensions of X.


```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)
```

b. Calculate: X.T and explain what the transpose represents.


```python
print("X transpose:")
print(X.T)
print("The transpose changes rows into columns and columns into rows.")
```

c. Calculate the matrix product: X.T @ X


```python
XTX = X.T @ X
print("X.T @ X:")
print(XTX)
```

d. Calculate the matrix inverse using: np.linalg.inv() where appropriate.


```python
XTX_inverse = np.linalg.inv(XTX)
print("Inverse of X.T @ X:")
print(XTX_inverse)
```

e. Implement the Ordinary Least Squares equation: β = (X.T X)^-1 X.T y using NumPy.


```python
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("Beta coefficients:", beta)
```

f. Explain what the resulting coefficients represent in the context of the three user features.


```python
print("The coefficients show how the predicted assistance score changes")
print("with sleep hours, activity level, and stress level.")
```

g. A new user has: new_user = np.array([5, 40, 7]) Use your calculated coefficients to obtain a predicted assistance score.


```python
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("Predicted assistance score:", predicted_score)
```
