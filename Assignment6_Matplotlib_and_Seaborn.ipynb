Q1. Ask the user to enter a value (e.g., M) for a mathematical function.


• Generate x values from -10 to 10 using np.linspace().


• Compute y values for:


y = M⋅x²


y = M⋅sin(x)


• Plot both functions in a single figure:


Use different colors and line styles.


Add legend, grid, and title.


```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

M = float(input("Enter value of M: "))
x = np.linspace(-10, 10, 100)
y1 = M * x**2
y2 = M * np.sin(x)
plt.plot(x, y1, label="M*x²", linestyle="-")
plt.plot(x, y2, label="M*sin(x)", linestyle="--")
plt.title("Mathematical Functions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
```

Q.2 Create a dataset of five subjects and scores.


• Convert it into a Pandas DataFrame.


• Plot the scores using a Seaborn bar plot with:


Different colors for each bar.


Annotations on top of each bar.


Title, axis labels, and grid.


```python
data = {
    "Subject": ["Maths", "Physics", "Chemistry", "English", "Computer"],
    "Score": [85, 78, 90, 82, 95]
}
df = pd.DataFrame(data)
print(df)
plt.figure(figsize=(8, 5))
ax = sns.barplot(data=df, x="Subject", y="Score", hue="Subject", legend=False)
for i in range(len(df)):
    plt.text(i, df["Score"][i] + 1, df["Score"][i], ha="center")
plt.title("Subject Scores")
plt.xlabel("Subject")
plt.ylabel("Score")
plt.grid(axis="y")
plt.show()
```

Q3. Write a Python script to select your roll number as seed for NumPy and then generate a dataset of 50 values using np.random.randn().


• Create a 2x2 subplot layout displaying:


Line plot showing cumulative sum.


Scatter plot with random noise.


• Customize with titles, axis labels, and grids.


```python
np.random.seed(1024170047)
data = np.random.randn(50)
cumulative_sum = np.cumsum(data)
noise = np.random.randn(50)
fig, ax = plt.subplots(2, 2, figsize=(10, 8))
ax[0, 0].plot(data)
ax[0, 0].set_title("Random Data")
ax[0, 0].set_xlabel("Index")
ax[0, 0].set_ylabel("Value")
ax[0, 0].grid()
ax[0, 1].plot(cumulative_sum)
ax[0, 1].set_title("Cumulative Sum")
ax[0, 1].set_xlabel("Index")
ax[0, 1].set_ylabel("Cumulative Value")
ax[0, 1].grid()
ax[1, 0].scatter(range(50), data + noise)
ax[1, 0].set_title("Scatter Plot with Random Noise")
ax[1, 0].set_xlabel("Index")
ax[1, 0].set_ylabel("Value")
ax[1, 0].grid()
ax[1, 1].plot(data)
ax[1, 1].set_title("Random Values")
ax[1, 1].set_xlabel("Index")
ax[1, 1].set_ylabel("Value")
ax[1, 1].grid()
plt.tight_layout()
plt.show()
```

Q.4 Download Data-set from the below link


https://github.com/AnjulaMehto/MCA/blob/main/company_sales_data.csv


Apply ‘seaborn’ library to do the following.


1. Read Total profit of all months and show it using a line plot.


```python
url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"
sales_data = pd.read_csv(url)
print(sales_data.head())
plt.figure(figsize=(8, 5))
sns.lineplot(data=sales_data, x="month_number", y="total_profit")
plt.title("Total Profit of All Months")
plt.xlabel("Month Number")
plt.ylabel("Total Profit")
plt.grid()
plt.show()
```

2. Read all product sales data and show it using a multiline plot.


```python
plt.figure(figsize=(10, 6))
for column in sales_data.columns:
    if column.endswith("_units"):
        sns.lineplot(data=sales_data, x="month_number", y=column, label=column)
plt.title("Product Sales Data")
plt.xlabel("Month Number")
plt.ylabel("Units Sold")
plt.legend()
plt.grid()
plt.show()
```

3. Plot bar chart for all the features/attributes.


```python
plt.figure(figsize=(12, 7))
for column in sales_data.columns:
    if column != "month_number":
        sns.barplot(data=sales_data, x="month_number", y=column)
plt.title("Company Sales Data")
plt.xlabel("Month Number")
plt.ylabel("Values")
plt.grid(axis="y")
plt.show()
```
