Q.2 Questions on Basic NumPy Array:


a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5])


```python
arr = np.array([1, 2, 3, 6, 4, 5])
print("Original array:", arr)
print("Reversed array:", arr[::-1])
```

b) Find the most frequent value and their indice(s) in the following arrays:


i. x = np.array([1,2,3,4,5,1,2,1,1,1])


```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
max_count = counts.max()
for value in values[counts == max_count]:
    print("Most frequent value:", value)
    print("Frequency:", max_count)
    print("Indices:", np.where(x == value)[0])
```

ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])


```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values, counts = np.unique(y, return_counts=True)
max_count = counts.max()
for value in values[counts == max_count]:
    print("Most frequent value:", value)
    print("Frequency:", max_count)
    print("Indices:", np.where(y == value)[0])
```

Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute?


```python
prabhleen = np.linspace(10, 100, 25)
print("Array:", prabhleen)
print("Dimensions:", prabhleen.ndim)
print("Shape:", prabhleen.shape)
print("Total elements:", prabhleen.size)
print("Data type:", prabhleen.dtype)
print("Total bytes:", prabhleen.nbytes)
transpose_array = prabhleen.reshape(1, 25)
print("Transpose using reshape():")
print(transpose_array)
print("Transpose using T:")
print(prabhleen.T)
print("Yes, T can be used, but for a 1-D array it does not change the shape.")
```

Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three columns and name it as resized_ ucs420_<your_name>>.


```python
ucs420_prabhleen = np.array([[10, 20, 30, 40],[50, 60, 70, 80],[90, 15, 20, 35]])
print("Array:")
print(ucs420_prabhleen)
print("Mean:", np.mean(ucs420_prabhleen))
print("Median:", np.median(ucs420_prabhleen))
print("Maximum:", np.max(ucs420_prabhleen))
print("Minimum:", np.min(ucs420_prabhleen))
print("Unique elements:", np.unique(ucs420_prabhleen))
reshaped_ucs420_prabhleen = ucs420_prabhleen.reshape(4, 3)
print("Reshaped array:")
print(reshaped_ucs420_prabhleen)
resized_ucs420_prabhleen = np.resize(ucs420_prabhleen, (2, 3))
print("Resized array:")
print(resized_ucs420_prabhleen)
```
